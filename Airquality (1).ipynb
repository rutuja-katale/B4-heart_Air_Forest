```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Air_Quality (1).csv")
```

```python
print(data.isnull().sum())
```

```python
data = data.drop_duplicates()
```

```python
plt.figure(figsize=(8,4))
data['City'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 States")
plt.xlabel("State")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(data['PM2.5'], data['CO'])

plt.title("PM2.5 vs CO")
plt.xlabel("PM2.5")
plt.ylabel("RSPM")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(data.corr(numeric_only=True),annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(7,7))

data['City'].value_counts().head(5).plot(kind='pie', autopct='%1.1f%%')

plt.title("Top 5 States")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.boxplot(y=data['SO2'])

plt.title("SO2 Box Plot")
plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.plot(data['O3'].head(100))

plt.title("O3 Line Plot")
plt.xlabel("Index")
plt.ylabel("O3")
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.histplot(data['PM2.5'], bins=20, kde=True)

plt.title("PM2.5 Distribution")
plt.xlabel("PM2.5")
plt.ylabel("Frequency")
plt.show()
```
