```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("forestfires.csv")
```

```python
print(data.head())
```

```python
print(data.columns)
```

```python
print("\nMissing Values:\n")
print(data.isnull().sum())
```

```python
data = data.drop_duplicates()
```

```python
plt.figure(figsize=(8,4))

sns.countplot(x='month', data=data)

plt.title("Forest Fires by Month")
plt.xlabel("Month")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(data['temp'], bins=20, kde=True)
plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='temp', y='area', data=data)
plt.title("Temperature vs Burned Area")
plt.xlabel("Temperature")
plt.ylabel("Burned Area")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(y='RH', data=data)
plt.title("Relative Humidity Box Plot")
plt.show()
```

```python
month_count = data['month'].value_counts().head(5)
plt.figure(figsize=(7,7))
plt.pie(
    month_count.values,
    labels=month_count.index,
    autopct='%1.1f%%'
)
plt.title("Top 5 Months with Forest Fires")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(data[['temp', 'RH', 'wind', 'rain', 'area']])
plt.show()
```
